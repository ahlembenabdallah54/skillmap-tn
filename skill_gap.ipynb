```python
import pandas as pd

offres = [
    {"category": "Data & BI", "skills": ["Python", "SQL", "Power BI"]},
    {"category": "Data & BI", "skills": ["SQL", "Machine Learning", "Python"]},
    {"category": "Data & BI", "skills": ["SQL", "Power BI", "Excel"]},
    {"category": "Data & BI", "skills": ["Python", "Machine Learning", "Pandas"]},
    {"category": "Data & BI", "skills": ["SQL", "Power BI", "Tableau"]},
    {"category": "Data & BI", "skills": ["Python", "SQL", "Machine Learning"]},
    {"category": "Data & BI", "skills": ["SQL", "Excel", "Power BI"]},
    {"category": "Data & BI", "skills": ["Python", "Deep Learning", "Machine Learning"]},
    {"category": "Data & BI", "skills": ["SQL", "Power BI", "Python"]},
    {"category": "Data & BI", "skills": ["Pandas", "SQL", "Tableau"]},
    {"category": "Développement logiciel", "skills": ["Java", "Spring"]},
    {"category": "Développement logiciel", "skills": ["JavaScript", "React"]},
]

profil = {"metier_cible": "Data & BI", "competences": ["Python", "SQL", "Pandas"]}
```

```python
def skill_gap(profil, offres, categorie):
    offres_cat = [o for o in offres if o["category"] == categorie]
    n = len(offres_cat)
    if n == 0:
        return "Données insuffisantes"
    comptes = {}
    for o in offres_cat:
        for s in set(o["skills"]):
            comptes[s] = comptes.get(s, 0) + 1
    manquantes = [
        (s, c, n, round(c / n * 100))
        for s, c in comptes.items()
        if s not in profil["competences"]
    ]
    return sorted(manquantes, key=lambda x: (-x[1], x[0]))
```

```python
resultat = skill_gap(profil, offres, profil["metier_cible"])
df = pd.DataFrame(resultat, columns=["Compétence", "Annonces", "Total", "Demande %"])
df.insert(0, "Priorité", range(1, len(df) + 1))
df
```
